### Remove Constant Variable

`application_type` contains only the `INDIVIDUAL` category across all observations. Since the variable provides no variation or discriminatory information, it is removed from the analytical dataset.

```python
loan_df.drop(columns="application_type", inplace=True)
```

### Missing value assessment

The `emp_title` variable contains 1,438 missing observations, representing approximately 3.7% of the dataset. It also contains a very high number of unique values, making it difficult to standardize and use effectively in the analysis.

Given its missingness and high cardinality, `emp_title` is removed from the dataset rather than imputing or grouping the values arbitrarily.

```python
missing = loan_df.isna().sum().sort_values(ascending=False)
missing = missing[missing > 0]

missing
```

```python
missing_pct = (loan_df.isna().mean() * 100).sort_values(ascending=False)
missing_pct = missing_pct[missing_pct > 0]

missing_pct
```

```python
loan_df.drop(columns="emp_title", inplace=True)
```

```python
loan_df.isnull().sum()
```

### Standardize Categorical Variables

Categorical values are standardized to maintain consistent formatting and prevent equivalent categories from being treated as separate values because of differences in whitespace or representation.

```python
loan_df.columns = (
    loan_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
categorical_columns = loan_df.select_dtypes(include="object").columns

for col in categorical_columns:
    loan_df[col] = loan_df[col].str.strip()
```

### Convert Loan Term

The `term` variable contains the loan duration as text, such as `36 months` and `60 months`.

The numerical duration is extracted and stored as `term_months` so that loan tenure can be analysed as a numerical variable.

```python
loan_df["term"] = loan_df["term"].str.replace(" months", "").astype(int)
loan_df.rename(columns={"term": "term_months"}, inplace=True)
```

### Convert Employment Length

`emp_length` represents employment tenure using text categories ranging from less than one year to ten or more years.

The categories are converted into numerical years to make employment tenure suitable for quantitative analysis and modelling.

```python
emp_length_mapping = {
    "< 1 year": 0,
    "1 year": 1,
    "2 years": 2,
    "3 years": 3,
    "4 years": 4,
    "5 years": 5,
    "6 years": 6,
    "7 years": 7,
    "8 years": 8,
    "9 years": 9,
    "10+ years": 10
}

loan_df["emp_length"] = loan_df["emp_length"].map(emp_length_mapping)
```

```python
print(loan_df["term_months"].unique())
print(loan_df["emp_length"].unique())
```

### Numerical Variable Assessment

Numerical variables are assessed to understand their ranges, distributions, and potential extreme observations before proceeding with further analysis.

The assessment includes:

- Reviewing the minimum and maximum values of the numerical variables.
- Calculating skewness to assess the shape and symmetry of the distributions.
- Examining selected percentiles to understand the spread and identify potential extreme values.
- Investigating observations with unusually high or low values alongside relevant loan characteristics.
- Assessing whether extreme observations provide sufficient evidence of potential data-quality issues.

Extreme values are not removed automatically. Any decision to retain or remove observations is based on further examination of the data rather than magnitude alone.

#### Examine Minimum and Maximum Values

Minimum and maximum values are examined to understand the range of each numerical variable (excluding `id` and `member_id`) and identify potentially unusual observations.

```python
numerical_columns = loan_df.select_dtypes(include=np.number).columns.tolist()

numerical_columns.remove("id")
numerical_columns.remove("member_id")
```

```python
loan_df[numerical_columns].min()
```

```python
loan_df[numerical_columns].max()
```

#### Assess Distribution and Skewness

Skewness is calculated to assess the symmetry of the numerical variables.

The results show substantial positive skewness in `annual_income`, while variables such as `dti` are comparatively symmetric.

```python
loan_df[numerical_columns].skew()
```

#### Examine Percentile Distributions

Selected percentiles were examined to understand the spread of the numerical variables and identify potential extreme values.

`annual_income` shows the most pronounced upper-tail behavior. While 99% of observations have an income below $235,000, the maximum reaches $6,000,000, which is consistent with its high positive skewness.

`loan_amount`, `installment`, `total_acc`, and `total_payment` also show right-skewed distributions, but their extreme values remain within plausible ranges.

At this stage, extreme observations are not removed automatically. Their validity will be assessed further during exploratory data analysis.

```python
loan_df[numerical_columns].quantile([0.01, 0.25, 0.50, 0.75, 0.95, 0.99])
```

#### Investigate Extreme Observations

Variables with unusually high skewness are investigated further by examining their extreme observations alongside relevant loan characteristics. Extreme values are not automatically removed because a large value is not necessarily an erroneous value.

The extreme observations are primarily concentrated in `annual_income`, with several borrowers reporting incomes substantially above the typical range. However, these observations are associated with plausible loan amounts, grades, DTI values, and loan outcomes, suggesting that they should not be removed solely because of their magnitude.

For example, the borrower with an annual income of $6,000,000 has a $5,000 loan and a DTI of 0.01, and the loan was fully paid. Other high-income observations also show varying loan characteristics and outcomes.

Therefore, these observations are retained at this stage. Further analysis during EDA will determine whether they represent legitimate high-income borrowers or potential data-quality issues.

```python
loan_df.nlargest(10, "annual_income")[["annual_income", "loan_amount", "dti", "grade", "loan_status"]]
```

**Insight**

The numerical variables were assessed using descriptive statistics, skewness, and percentile distributions to identify potential data-quality issues and extreme observations.

`annual_income` showed substantial right skewness (31.07), with a maximum value of $6,000,000 compared with a median of $60,000. The extreme observations were investigated alongside loan amount, DTI, grade, and loan status. The values appeared plausible and did not provide sufficient evidence of data-entry errors.

The remaining numerical variables showed moderate or low skewness and plausible minimum and maximum values.

**Decision:** No numerical observations were removed or capped at this stage. Extreme values in `annual_income` were retained as potentially legitimate borrower characteristics. Any transformation required to address skewness will be considered during exploratory analysis and/or model preparation.
